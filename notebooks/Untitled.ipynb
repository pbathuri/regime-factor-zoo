```python
# Cell 1 — Ken French Fama-French 3-factor, monthly
import pandas_datareader.data as pdr
import pandas as pd

ff3 = pdr.DataReader("F-F_Research_Data_Factors", "famafrench",
                     start="1926-07-01")[0]
ff3.index = ff3.index.to_timestamp()
print(ff3.shape)
print(ff3.tail())
ff3.to_parquet("data/processed/ff3_monthly.parquet")
```

```python
# Cell 2 — Momentum (Carhart)
mom = pdr.DataReader("F-F_Momentum_Factor", "famafrench", start="1926-11-01")[0]
mom.index = mom.index.to_timestamp()
mom.to_parquet("data/processed/momentum_monthly.parquet")
print(mom.tail())
```

```python
# Cell 3 — yfinance SPY sanity check
import yfinance as yf
spy = yf.download("SPY", start="1993-01-29", auto_adjust=False, progress=False)
print(spy.shape)
spy.to_parquet("data/raw/spy_daily.parquet")
print(spy.tail())
```

```python
# Cell 4 — Cumulative SPY plot
import matplotlib.pyplot as plt
spy["ret"] = spy["Adj Close"].pct_change()
(1 + spy["ret"].dropna()).cumprod().plot(logy=True, figsize=(10,5),
    title="SPY cumulative return (yfinance)")
plt.tight_layout()
plt.savefig("figures/01_spy_cumulative.png", dpi=120, bbox_inches="tight")
plt.show()
```
